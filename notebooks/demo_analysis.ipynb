# Synchronization Analysis - Demo Notebook

This notebook demonstrates the unified synchronization analysis framework
for different types of time series data.

```python
# Setup
import sys
sys.path.insert(0, '../src')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

from synch_analysis import (
    LorenzDataSource,
    SinusoidDataSource,
    CoupledOscillatorDataSource,
    ParquetDataSource,
    SynchronizationPipeline,
    compare_data_sources,
    export_results,
)
```

## 1. Lorenz Attractor Analysis

Two Lorenz systems with slightly different initial conditions.

```python
# Create Lorenz data source
lorenz = LorenzDataSource(
    a=10.0, b=28.0, c=8.0/3.0, dt=0.01,
    initial_values_1=[0.01, 0, 0.3],
    initial_values_2=[0.2, 0.1, 0.4],
    iterations=1000,
    variable='x',
    sampling_rate=100.0
)

# Run pipeline
lorenz_pipeline = SynchronizationPipeline(lorenz).run()

# Show dashboard
fig = lorenz_pipeline.plot_dashboard(sliding_window=30, n_signal_samples=300)
plt.show()
```

## 2. Sinusoidal Signals

Two sinusoids with a phase delay.

```python
# Create sinusoid data source
sinusoid = SinusoidDataSource(
    ph0_a=0, ph0_b=0,
    frq_a=2, frq_b=2,
    pers=10, delay_a=0, delay_b=3,
    iterations=1000,
    sampling_rate=1.0
)

# Run pipeline
sin_pipeline = SynchronizationPipeline(sinusoid).run()

# Show dashboard
fig = sin_pipeline.plot_dashboard(sliding_window=30, n_signal_samples=300)
plt.show()
```

## 3. Coupled Oscillators (Kuramoto Model)

Synthetic oscillators with controllable coupling strength.

```python
# Create coupled oscillator data source
coupled = CoupledOscillatorDataSource(
    n_oscillators=2,
    coupling_strength=0.8,
    natural_freqs=np.array([1.0, 1.05]),
    dt=0.01,
    duration=50.0,
    noise_std=0.01,
    sampling_rate=100.0
)

# Run pipeline
coupled_pipeline = SynchronizationPipeline(coupled).run()

# Show dashboard
fig = coupled_pipeline.plot_dashboard(sliding_window=50, n_signal_samples=500)
plt.show()
```

## 4. Batch Comparison

Compare synchronization across different data sources.

```python
# Compare multiple sources
sources = [
    LorenzDataSource(iterations=500),
    SinusoidDataSource(frq_a=2, frq_b=2, delay_a=0, delay_b=3, iterations=500),
    CoupledOscillatorDataSource(coupling_strength=0.8, natural_freqs=np.array([1.0, 1.05]), duration=20.0),
]
labels = ['Lorenz', 'Sinusoids', 'Coupled Oscillators']

comparison_df = compare_data_sources(sources, labels)
display(comparison_df)
```

## 5. Export Results

Save analysis results to files.

```python
# Export results for coupled oscillators
export_results(coupled_pipeline, output_dir='../results/coupled_oscillators')
```
